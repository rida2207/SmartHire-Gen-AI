```python
from sentence_transformers import SentenceTransformer

print("Loading Sentence Transformer model directly...")
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
sample_texts = [
    "Experienced Python developer with a background in machine learning and data engineering.",
    "Looking for a junior software engineering role focused on web development."
]

print("\nGenerating embeddings for sample texts...")
vectors = model.encode(sample_texts)

print(f"Successfully generated {len(vectors)} embedding vectors!")
print(f"Dimension of each vector: {len(vectors[0])}")
```
